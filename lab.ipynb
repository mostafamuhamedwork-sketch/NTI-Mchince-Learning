### Question: What is the result of the following expressions?
print(0 * np.nan)

print(np.nan == np.nan)

print(np.inf > np.nan)

print(np.nan - np.nan)

print(np.nan in set([np.nan]))

print(0.3 == 3 * 0.1)

### Answer

```python
# import numpy as np

# # 1. أي عملية حسابية مع np.nan تعطي دائماً nan
# print(0 * np.nan)                  # الناتج: nan

# # 2. القيمة nan لا تساوي نفسها أبداً حسب معايير IEEE 754
# print(np.nan == np.nan)            # الناتج: False

# # 3. أي مقارنة (أكبر من، أصغر من، تساوي) تتضمن nan ترجع False
# print(np.inf > np.nan)             # الناتج: False

# # 4. طرح nan من nan يظل الناتج nan
# print(np.nan - np.nan)             # الناتج: nan

# # 5. الـ set تتعرف على وجود nan لأنها تعتمد على الـ hashing
# print(np.nan in set([np.nan]))     # الناتج: True

# # 6. مشكلة دقة تمثيل الأرقام العشرية (Floating-point precision)
# print(0.3 == 3 * 0.1)              # الناتج: False
```

```python
print(0 * np.nan) #NaN تعني "قيمة غير معرفة أو غير صالحة"

print(np.nan == np.nan)

print(np.inf > np.nan)
# np.inf تعني ما لا نهاية (Infinity)

print(np.nan - np.nan)

print(np.nan in set([np.nan])) # دى  set([np.nan]) بتتعرف على ال nan

print(0.3 == 3 * 0.1)
```

### Question: Write a NumPy program to sort pairs of first name and last name return their indices. (first by last name, then by first name).
first_names = (‘Betsey’, ‘Shelley’, ‘Lanell’, ‘Genesis’, ‘Margery’)

last_names = (‘Battle’, ‘Brien’, ‘Plotner’, ‘Stahl’, ‘Woolum’)

### Answer:

```python


first_names = ('Betsey', 'Shelley', 'Lanell', 'Genesis', 'Margery')
last_names = ('Battle', 'Brien', 'Plotner', 'Stahl', 'Woolum')

indices = np.lexsort((first_names, last_names))

print(indices)
```

### Question: How to tell if a given 2D array has null columns?

### Answer:

```python
arr = np.array([
    [1, np.nan, 3],
    [4, np.nan, 6],
    [7, np.nan, 9]
])
null_columns = np.all(np.isnan(arr), axis=0)
print(null_columns)
print("Has null columns:", np.any(null_columns))
```
